Read Excel File

```python
import pandas as pd 
df = pd.read_excel("C:\\Users\\Rc\\Downloads\\Pandas\\stock_data.xlsx","Sheet1")
```

```python
df
```

```python
def convert_people_cell(cell):
    if cell=="n.a.":
        return 'Sam Walton'
    return cell

def convert_price_cell(cell):
    if cell=="n.a.":
        return 50
    return cell
    
df = pd.read_excel("C:\\Users\\Rc\\Downloads\\Pandas\\stock_data.xlsx","Sheet1", converters= {
        'people': convert_people_cell,
        'price': convert_price_cell
    })
df
```

Write to Excel

```python
df.to_excel("C:\\Users\\Rc\\Downloads\\Pandas\\new_stock_4.xlsx", sheet_name="stocks", index=False, startrow=2, startcol=1)
```
