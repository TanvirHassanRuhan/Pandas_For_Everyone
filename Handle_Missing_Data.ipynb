```python
import pandas as pd 
df =  pd.read_csv("C:\\Users\\Rc\\Downloads\\Pandas\\weather_data _with_missing_value.csv",parse_dates=["day"])
df.set_index('day',inplace = True)
```

```python
df
```

# fillna() Function

```python
new_df = df.fillna(0)
new_df
```

```python
new_df = df.fillna({
    'temperature':0,
    'windspeed':0,
    'event': 'no event'})
new_df
```

```python
new_df = df.fillna(method = "ffill")
new_df
```

```python
new_df = df.fillna(method = "bfill")
new_df
```

```python
new_df
```

# interpolate()

```python
new_df = df.interpolate()
```

```python
new_df
```

```python
new_df = df.interpolate(method = "time")
new_df
```

# dropna()

```python
new_df = df.dropna()
new_df
```

```python
new_df = df.dropna(how = "all")
new_df
```

```python
new_df = df.dropna(thresh = 1)
new_df
```

```python
new_df = df.dropna(thresh = 2)
new_df
```
